```python
students = [
    {"name": "Mike", "age": 17, "score": 80},
    {"name": "Sarah", "age": 20, "score": 45},
    {"name": "John", "age": 22, "score": 60},
    {"name": "Anna", "age": 18, "score": 55},
    {"name": "Tom", "age": 16, "score": 90}
]

filtered = [
    {
        "student_name" : student["name"],
        "age" : student["age"],
        "score" : student["score"],
        "status" : "Pass" if student["score"] >= 60 else "Fail"
    }
    for student in students if student["age"] >= 18 and student['score'] >= 60
]

for i in filtered:
    print(i)
```
